```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("/content/1776311302-P3-Car Market Trends Analysis with Car Dekho Data.csv")

data.head()
```

```python
data['Fuel_Type'].value_counts()
```

```python
import matplotlib.pyplot as plt

data['Fuel_Type'].value_counts().plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Distribution of Cars by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Number of Cars')
plt.xticks(rotation=0)
plt.show()
```

```python
plt.figure(figsize=(10, 5))

plt.hist(data['Selling_Price'], bins=30)

plt.title('Distribution of Car Selling Prices')
plt.xlabel('Selling Price (Lakhs)')
plt.ylabel('Number of Cars')

plt.show()
```

```python
avg_price = data.groupby('Fuel_Type')['Selling_Price'].mean()

print(avg_price)
```

```python
avg_price = data.groupby('Fuel_Type')['Selling_Price'].mean()

avg_price.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Selling Price by Fuel Type')
plt.xlabel('Fuel Type')
plt.ylabel('Average Selling Price (Lakhs)')
plt.xticks(rotation=0)

plt.show()
```

```python
top_cars = data['Car_Name'].value_counts().head(10)

print(top_cars)
```

```python
top_cars = data['Car_Name'].value_counts().head(10)

top_cars.plot(
    kind='bar',
    figsize=(10,6)
)

plt.title('Top 10 Most Common Car Models')
plt.xlabel('Car Model')
plt.ylabel('Number of Cars')
plt.xticks(rotation=45, ha='right')

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(data['Year'], data['Selling_Price'])

plt.title('Year vs Selling Price')
plt.xlabel('Manufacturing Year')
plt.ylabel('Selling Price (Lakhs)')

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(data['Kms_Driven'], data['Selling_Price'])

plt.title('Kms Driven vs Selling Price')
plt.xlabel('Kilometers Driven')
plt.ylabel('Selling Price (Lakhs)')

plt.show()
```

```python
plt.figure(figsize=(10,6))

plt.scatter(data['Present_Price'], data['Selling_Price'])

plt.title('Present Price vs Selling Price')
plt.xlabel('Present Price (Lakhs)')
plt.ylabel('Selling Price (Lakhs)')

plt.show()
```

```python
owner_price = data.groupby('Owner')['Selling_Price'].mean()

owner_price.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Selling Price by Number of Owners')
plt.xlabel('Number of Previous Owners')
plt.ylabel('Average Selling Price (Lakhs)')
plt.xticks(rotation=0)

plt.show()
```

```python
trans_price = data.groupby('Transmission')['Selling_Price'].mean()

trans_price.plot(
    kind='bar',
    figsize=(8,5)
)

plt.title('Average Selling Price by Transmission Type')
plt.xlabel('Transmission Type')
plt.ylabel('Average Selling Price (Lakhs)')
plt.xticks(rotation=0)

plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

correlation = data[['Year', 'Selling_Price', 'Present_Price', 'Kms_Driven', 'Owner']].corr()

plt.figure(figsize=(8,6))

sns.heatmap(correlation, annot=True, cmap='coolwarm', fmt='.2f')

plt.title('Correlation Between Car Attributes')

plt.show()
```

```python
print("Total Cars:", len(data))
print("Unique Car Models:", data['Car_Name'].nunique())

print("\nMost Common Fuel Type:")
print(data['Fuel_Type'].value_counts().idxmax())

print("\nMost Common Transmission:")
print(data['Transmission'].value_counts().idxmax())

print("\nMost Common Seller Type:")
print(data['Seller_Type'].value_counts().idxmax())

print("\nAverage Selling Price:")
print(round(data['Selling_Price'].mean(), 2), "Lakhs")

print("\nAverage Present Price:")
print(round(data['Present_Price'].mean(), 2), "Lakhs")

print("\nAverage Kms Driven:")
print(round(data['Kms_Driven'].mean(), 0))

print("\nCorrelation with Selling Price:")
print(
    data[['Year', 'Selling_Price', 'Present_Price', 'Kms_Driven', 'Owner']]
    .corr()['Selling_Price']
    .sort_values(ascending=False)
)
```
